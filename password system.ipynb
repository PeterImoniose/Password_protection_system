#### Multi-user account

```python
from getpass import getpass

class AccountSystem:
    def __init__(self):
        self.users = {}  # Store users and their passwords

    def create_account(self):
        username = input("Enter a username: ").strip()
        if username in self.users:
            print("Username already exists! Try another.")
            return

        pin = getpass("Create a 4-digit password: ")
        if not pin.isdigit() or len(pin) != 4:
            print("Password must be exactly 4 digits.")
            return

        self.users[username] = int(pin)
        print("ACCOUNT CREATED SUCCESSFULLY!")

    def login(self):
        username = input("Enter your username: ").strip()
        if username not in self.users:
            print("No such user. Please create an account first.")
            return

        try:
            pin_parts = input("Enter your 4-digit PIN separated by commas (e.g. 1,2,3,4): ")
            p1, p2, p3, p4 = map(int, pin_parts.split(","))
            entered_pin = int(f"{p1}{p2}{p3}{p4}")
        except ValueError:
            print("Invalid format. Please enter 4 numbers separated by commas.")
            return

        if self.users[username] == entered_pin:
            print("Login successful! Welcome,", username)
        else:
            print("Incorrect password!")

    def logout(self):
        print("Logged out successfully.")

def main():
    system = AccountSystem()
    while True:
        print("\n<<< WELCOME TO THE ACCOUNT SYSTEM >>>")
        print("1. Create Account")
        print("2. Login")
        print("3. Exit")

        choice = input("Choose an option: ")
        if choice == "1":
            system.create_account()
        elif choice == "2":
            system.login()
        elif choice == "3":
            print("Thank you for using the system. Goodbye!")
            break
        else:
            print("Invalid choice. Please select a valid option.")

if __name__ == "__main__":
    main()
```

#### Lockout System

```python
from getpass import getpass

class AccountSystem:
    def __init__(self):
        self.users = {}        
        self.attempts = {}    
        self.locked_users = set()

    def create_account(self):
        username = input("Enter a username: ").strip()
        if username in self.users:
            print("Username already exists! Try another.")
            return

        pin = getpass("Create a 4-digit password: ")
        if not pin.isdigit() or len(pin) != 4:
            print("Password must be exactly 4 digits.")
            return

        self.users[username] = int(pin)
        self.attempts[username] = 0
        print("ACCOUNT CREATED SUCCESSFULLY!")

    def login(self):
        username = input("Enter your username: ").strip()

        if username not in self.users:
            print("No such user. Please create an account first.")
            return

        if username in self.locked_users:
            print("This account is locked due to too many failed attempts.")
            return

        try:
            pin_parts = input("Enter your 4-digit PIN separated by commas (e.g. 1,2,3,4): ")
            p1, p2, p3, p4 = map(int, pin_parts.split(","))
            entered_pin = int(f"{p1}{p2}{p3}{p4}")
        except ValueError:
            print("Invalid input format. Use 4 digits separated by commas.")
            return

        if self.users[username] == entered_pin:
            print(f"Login successful! Welcome, {username}")
            self.attempts[username] = 0  # reset attempts on success
        else:
            self.attempts[username] += 1
            remaining = 3 - self.attempts[username]
            print(f"Incorrect password! Attempts left: {remaining}")
            if self.attempts[username] >= 3:
                self.locked_users.add(username)
                print("Account locked due to too many failed attempts.")

    def logout(self):
        print("Logged out successfully.")

def main():
    system = AccountSystem()
    while True:
        print("\n<<< ACCOUNT SYSTEM >>>")
        print("1. Create Account")
        print("2. Login")
        print("3. Exit")

        choice = input("Choose an option: ")
        if choice == "1":
            system.create_account()
        elif choice == "2":
            system.login()
        elif choice == "3":
            print("Thank you for using the system. Goodbye!")
            break
        else:
            print("Invalid choice. Please select a valid option.")

if __name__ == "__main__":
    main()
```

```python
from getpass import getpass

class AccountSystem:
    def __init__(self):
        self.users = {}             # e.g., {"dave": {"pin": 1234, "question": "pet?", "answer": "tiger"}}
        self.attempts = {}
        self.locked_users = set()

    def create_account(self):
        username = input("Enter a username: ").strip()
        if username in self.users:
            print("Username already exists!")
            return

        pin = getpass("Create a 4-digit PIN: ")
        if not pin.isdigit() or len(pin) != 4:
            print("PIN must be exactly 4 digits.")
            return

        question = input("Set a secret question for password recovery: ")
        answer = input("Answer to your secret question: ").strip().lower()

        self.users[username] = {
            "pin": int(pin),
            "question": question,
            "answer": answer
        }
        self.attempts[username] = 0
        print("Account created successfully!")

    def login(self):
        username = input("Enter your username: ").strip()
        if username not in self.users:
            print("No such user.")
            return

        if username in self.locked_users:
            print("Account is locked due to too many failed attempts.")
            return

        try:
            pin_input = input("Enter your 4-digit PIN separated by commas (e.g. 1,2,3,4): ")
            p1, p2, p3, p4 = map(int, pin_input.split(","))
            entered_pin = int(f"{p1}{p2}{p3}{p4}")
        except ValueError:
            print("Invalid input. Use 4 digits separated by commas.")
            return

        if entered_pin == self.users[username]["pin"]:
            print(f"Welcome, {username}! Login successful.")
            self.attempts[username] = 0
        else:
            self.attempts[username] += 1
            remaining = 3 - self.attempts[username]
            print(f"Incorrect PIN. Attempts left: {remaining}")
            if self.attempts[username] >= 3:
                self.locked_users.add(username)
                print("Account locked due to too many failed attempts.")

    def reset_password(self):
        username = input("Enter your username to reset password: ").strip()
        if username not in self.users:
            print("User not found.")
            return
        if username not in self.locked_users:
            print("Account is not locked. Try logging in instead.")
            return

        print(f"Answer this to reset your password: {self.users[username]['question']}")
        answer = input("Your answer: ").strip().lower()
        if answer == self.users[username]["answer"]:
            new_pin = getpass("Enter a new 4-digit PIN: ")
            if not new_pin.isdigit() or len(new_pin) != 4:
                print("PIN must be 4 digits.")
                return
            self.users[username]["pin"] = int(new_pin)
            self.attempts[username] = 0
            self.locked_users.remove(username)
            print("Password reset successfully. You can now log in.")
        else:
            print("Incorrect answer. Cannot reset password.")

    def logout(self):
        print("Logged out successfully.")

def main():
    system = AccountSystem()
    while True:
        print("\n<<< ACCOUNT SYSTEM >>>")
        print("1. Create Account")
        print("2. Login")
        print("3. Reset Password")
        print("4. Exit")

        choice = input("Choose an option: ")
        if choice == "1":
            system.create_account()
        elif choice == "2":
            system.login()
        elif choice == "3":
            system.reset_password()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid option. Try again.")

if __name__ == "__main__":
    main()
```
